```python
import json

with open('C:/Proyectos/Proyecto_TTAD_1/Proyecto_TTAD_1/data.json', 'r', encoding='utf-8') as archivo:
  datos = json.load(archivo)

print(datos)
```

```python
# 1.producto con más ingresos

ingresos_por_producto = {}

for venta in datos['ventas']:
    producto = venta['producto']
    subtotal = venta['cantidad'] * venta['precio_unitario']
    ingresos_por_producto[producto] = (
        ingresos_por_producto.get(producto, 0.0) + subtotal
    )

# Obtener el producto con mayor facturación total
producto_mas_ingresos = max(
    ingresos_por_producto, key=ingresos_por_producto.get
)
total_ingresos = ingresos_por_producto[producto_mas_ingresos]

print(
    f'Producto con más ingresos: {producto_mas_ingresos} ({total_ingresos:.2f} €)'
)
```

```python
# 2.filtra las ventas por un vendedor concreto

vendedor_buscado = 'Ana'

# Filtrar las ventas del vendedor seleccionado
ventas_vendedor = [
    venta for venta in datos['ventas'] if venta['vendedor'] == vendedor_buscado
]

print(f'Ventas de {vendedor_buscado} ({len(ventas_vendedor)} registros):')
for v in ventas_vendedor:
    subtotal = v['cantidad'] * v['precio_unitario']
    print(
        f"- ID {v['id']} | {v['mes']}: {v['cantidad']}x {v['producto']}"
        f' ({subtotal:.2f} €)'
    )
```

```python
# 3.calcula el ticket medio
ventas = datos['ventas']
total_facturado = sum(
    venta['cantidad'] * venta['precio_unitario'] for venta in ventas
)
numero_ventas = len(ventas)

# Evitar división por cero si la lista estuviera vacía
ticket_medio = total_facturado / numero_ventas if numero_ventas > 0 else 0.0

print(f'Total facturado: {total_facturado:.2f} €')
print(f'Número de ventas: {numero_ventas}')
print(f'Ticket medio: {ticket_medio:.2f} €')
```
